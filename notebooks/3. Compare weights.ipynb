```python
from pathlib import Path
import pandas as pd
path_weights = Path('./../data/sample_weights.xlsx').resolve()
```

```python
sample_weights = pd.read_excel(path_weights, index_col=0)
pct_diff = (sample_weights - sample_weights.loc["T0"]) / sample_weights.loc["T0"] * 100
pct_diff["mean"] = pct_diff.mean(axis=1)
pct_diff["std"] = pct_diff.std(axis=1)
print(pct_diff.round(3))
```
