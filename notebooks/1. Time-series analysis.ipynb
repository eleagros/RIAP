```python
%load_ext autoreload
%autoreload 2
from pathlib import Path
from riap.riap import Riap
```

### 1. Initiate the analyzer

```python
alignment_method = "elastix"
data_path = Path('./../data/measurements/')
output_path = Path('./../output/impv1') / alignment_method
output_path.mkdir(exist_ok = True, parents = True)

riap_analyser = Riap(
    data_path = str(data_path.resolve()),           # folder containing the data
    path_output = str(output_path.resolve()),       # output folder for the results
    time_base = 'T0',                               # the identifier of the reference image (must be contained in the file name)
    PDDN = False,                                   # keep False
    metric = 'median',                              # the metric to use ("mean", "median" or "max")
    alignment_method = alignment_method,            # elastix or superglue (default is superglue, elastix requires matlab integration)
    instrument = 'IMPV1'                            # IMPV1 or IMPV2 - depending on the instrument used - slight changes in the paths
)
```

### 2. Perform the alignment (WARNING: THIS ERASES PREVIOUS COMPUTATIONS)

```python
riap_analyser.process()
```

### 3. Generates the output for plotting

```python
riap_analyser.compare_parameters()
```
