```python
%load_ext autoreload
%autoreload 2
from pathlib import Path
from riap.riap import Riap
```

### 1. Initiate the analyzer

```python
data_path = Path('./../data/measurements')
output_path = Path('./../output')
output_path.mkdir(exist_ok = True)

riap_analyser = Riap(
    data_path = str(data_path.resolve()),           # folder containing the data
    path_output = str(output_path.resolve()),       # output folder for the results
    time_base = 'T0',                               # the identifier of the reference image (must be contained in the file name)
    PDDN = False,                                   # keep False
    metric = 'median',                              # the metric to use ("mean", "median" or "max")
    alignment_method = 'superglue'                  # elastix or superglue (default is superglue, elastix requires matlab integration - see matlab.engine https://www.mathworks.com/help/matlab/matlab_external/install-the-matlab-engine-for-python.html if you want to use it)
)
```

### 2. Perform the alignment (WARNING: THIS ERASES PREVIOUS COMPUTATIONS)
